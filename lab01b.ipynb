```python
#input date
rawdate=input("Enter the date of the week you want to calculate. ex)Monday")

#date validation
valid_days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")

date = rawdate.capitalize()
if date not in valid_days:
    print("Invalid date. Please enter a valid date of the week like Monday, Tuesday, etc.")
    raise SystemExit(1)

#input order amount
rawfee=input("Enter your order amount to calculate the delivery fee. ex) 58")

#order amount validation
MIN_o=30
try:
    fee=float(rawfee)
except:
    print("Invalid fee format. Please enter a valid number.")
    raise SystemExit(1)
if fee < MIN_o:
    print(f"Order amount is less than the minimum required CAD{MIN_o}.")
    raise SystemExit(1)

#dynamic delivery fee
if date in ("Monday", "Tuesday"):
    delivery_fee = 5
elif date in ("Wednesday", "Thursday"): 
    delivery_fee = 3
elif date in ("Friday", "Saturday"):
    delivery_fee = 2
else:
    delivery_fee = 0

#delivery fee calculation
print(f"Your order amount is {fee}. The delivery fee for {date} is {delivery_fee}. \nYour total amount to pay is {fee + delivery_fee}.")
```
